NOTE ___________________________________________________________


If we do not specify the points on the x-axis, they will get the default values 0, 1, 2, 3 etc., depending on the length of the y-points.

```python
#Using Markers 

x = np.array([1,2,3,4,5])
y = np.array([250,389,238,500,100])

plt.plot(x,y, marker = "o:r")
plt.show()
```

```python
# marker = *
x = np.array([1,2,3,4,5])
y = np.array([250,389,238,500,100])

plt.plot(x,y, marker = "*")
plt.show()
```

```python
# marker|line|color
x = np.array([1,2,3,4,5])
y = np.array([250,389,238,500,100])

plt.plot(x,y, '*:r') # Marker Line Colour
plt.show()
```

```python
#marker => eg. 'o'
#Marker size => ms = eg. 20
#marker-edge-color => mec = eg. 'r' 
#marker-face-color => mfc = eg. 'g'


#linestyle = "dotted" , "dashed"
#linewidth = eg. 20.5 , 10
```

Practice Q2 - Draw two lines by specifying a plt.plot() function for each line:

```python
#TITLE - - Giving a title to the graph
#Implementing Labels - Y axis and X axis

x = np.array([1,2,3,4,5])
y = np.array([250,389,238,500,100])

xp = np.array([1,3,5,8,10,])
yp = np.array([100,200,600,800,280])

x1 = np.array([2,4,6,8,10])
y1 = np.array([125,235,370,540])

plt.title("Brajeswari")
plt.xlabel("X axis ---->")

plt.ylabel("Y axis ------>")

plt.plot(x, y, xp, yp, x1[:len(y1)], y1)
plt.show()



```

Set Font Properties for Title and Labels


You can use the fontdict parameter in xlabel(), ylabel(), and title() to set font properties for the title and labels.
