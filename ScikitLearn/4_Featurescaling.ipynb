4.1 Why is scaling necessary?

Imagine we have the following customer:

            Feature	                Value
            
            Age	                    25
            Salary	                80,000

The salary is numerically much larger than the age.

Some algorithms use distances, dot products, or optimization procedures that are sensitive to differences in feature scales.

For example, K-Nearest Neighbors computes distances between observations.

Without scaling, salary differences could dominate the distance calculation even if age is also important.

Feature scaling brings numerical features onto comparable scales.
4.2 StandardScaler
StandardScaler transforms each numerical feature using:
\[
z=\frac{x-\mu}{\sigma}
\]

Where:
- \(x\) = Original value
- \(\mu\) = Mean of the feature
- \(\sigma\) = Standard deviation of the feature
Suppose study hours are:

[2, 4, 6]

The mean is 4, and the population standard deviation is approximately 1.633.

After standardization:
Original	Standardized
2	-1.225
4	0
6	1.225
