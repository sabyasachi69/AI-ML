IMPUTATION
Many machine learning model can not deal with the missing values themselves,so we need to replace that missing values with estimated values .

3 common ways to deal with MISSING VALUES

Mean _______________________________

Replace the missing value with the arithmetic average of the available values.

Mean=(20000+25000+30000+35000)/4 = 27500

Median______________________________________
Replace the missing value with the middle value of the sorted available observations.

2 middle values sum / 2 = median

MODE___________________________________________________

Replace the missing value with the most frequently occurring value. In Scikit-learn, the corresponding strategy is called most_frequent.


WHEN TO CHOOSE WHAT ? 

The practical rule is:
- Mean: Often reasonable for numerical data without extreme outliers.
- Median: Often more robust for skewed numerical data or data containing outliers.
- Most frequent: Useful for categorical columns, such as City or Gender.

```python
from sklearn.impute import SimpleImputer

age = np.array([agedata])
print(age)
```

Method	            Meaning

3. ENCODING CATEGORICAL DATA

Look at the City column:
Delhi
Mumbai
Pune


These are categories, not numerical values.

Many machine learning algorithms expect numerical input.

Therefore, we need to convert categories into numbers.

This process is called encoding.

```python
#For example 
y = ["No", "Yes", "No", "Yes"]
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
Encodery = LabelEncoder().fit_transform(y)
encoded_y = encoder.fit_transform(y)   #2
print(encoded_y,Encodery)

print("Classes:", encoder.classes_)
print("Encoded values:", encoded_y)
```

```python
# Using onehot encoding method
from sklearn.preprocessing import OneHotEncoder

encoder1 = OneHotEncoder(sparse_output=False)
encoded_city = encoder1.fit_transform(citydata)


```

```python
print(encoder1.get_feature_names_out())

print(encoded_city)
```

WHY ONE COLUMN ?? 

What about genuinely ordered categories?


Some categorical values actually have an order.

For example:

Low < Medium < High

For these, OrdinalEncoder can be appropriate.
Priority	Number
Low	            0
Medium	        1
High	        2

So remember:

- LabelEncoder → Target labels (y).
- OneHotEncoder → Input categories without a natural order.
- OrdinalEncoder → Input categories with a meaningful order.
